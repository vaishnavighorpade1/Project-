```python
y = data["Price"]
x = data[[ "Processor_Speed","RAM_Size","Storage_Capacity"]]
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.2)
```

```python
len(x_train)
```

```python
len(y_train)
```

```python
len(x_test)
```

```python
len(y_test)
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rfrmodel = RandomForestRegressor()
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_gridrfr = {"max_depth" : [2,5,10,15,20,25],
                 "max_features" : ["auto", "log2"] ,
                 "n_estimators" : [2,4,6,8,10,12]}
```

```python
gridrfr = GridSearchCV(rfrmodel , param_gridrfr)
```

```python
gridrfr.fit(x_train , y_train)
```

```python
gridrfr.best_params_
```

```python
predictions = gridrfr.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error
```

```python
mean_absolute_error(y_test , predictions)
```

```python
import joblib 
joblib.dump(gridrfr, "rf_model.pkl")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Predict using the best model from GridSearchCV
y_pred = gridrfr.best_estimator_.predict(x_test)

# Evaluate accuracy
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("📊 Model Performance:")
print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"R² Score: {r2:.4f}")
```
